## Ejercicio 2: Escalamiento de tickets de soporte técnico

### Ficha PEAS
* **P (Performance / Desempeño):** Maximizar la resolución oportuna de tickets críticos y la satisfacción de clientes clave, minimizando tiempos de espera excesivos.
* **E (Environment / Entorno):** Sistema de mesa de ayuda (Helpdesk) y cola de atención de tickets de soporte.
* **A (Actuators / Actuadores):** Asignar el ticket al nivel correspondiente ("escalar a Nivel 1", "escalar a Nivel 2" o "escalar a Nivel 3").
* **S (Sensors / Sensores):** Datos de entrada: `tiempo_espera_minutos` (entero), `nivel_urgencia` ("baja", "media", "alta") y `cliente_premium` (booleano).

### Justificación de las reglas
Las reglas priorizan la retención de clientes estratégicos y la atención rápida de incidencias severas. Cualquier requerimiento de un cliente premium o con urgencia alta se escala de inmediato a Nivel 3 para garantizar atención prioritaria. Incidencias de urgencia media o solicitudes con tiempos de espera superiores a 30 minutos se derivan a Nivel 2 para evitar cuellos de botella e insatisfacción. Los tickets de urgencia baja y poco tiempo de espera se mantienen en Nivel 1 para optimizar la carga operativa.

### Código

```python
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):
    if cliente_premium or nivel_urgencia == "alta":
        return "escalar a Nivel 3", "atencion prioritaria por cliente premium o alta urgencia"
    elif nivel_urgencia == "media" or tiempo_espera_minutos > 30:
        return "escalar a Nivel 2", "requiere nivel intermedio por urgencia media o espera prolongada (>30 min)"
    else:
        return "escalar a Nivel 1", "atencion estandar para urgencia baja con espera aceptable"
```

### Simulación y pruebas

```python
casos_prueba = [
    (10, "baja", False),   # Espera baja, urgencia baja, no premium -> Nivel 1
    (45, "baja", False),   # Espera alta (>30 min), urgencia baja, no premium -> Nivel 2
    (15, "media", False),  # Espera baja, urgencia media, no premium -> Nivel 2
    (5, "alta", False),    # Espera baja, urgencia alta, no premium -> Nivel 3
    (10, "baja", True),    # Urgencia baja pero cliente premium -> Nivel 3
    (35, "media", True)    # Espera alta, urgencia media y cliente premium -> Nivel 3
]

for espera, urgencia, premium in casos_prueba:
    accion, motivo = agente_soporte(espera, urgencia, premium)
    print(f"Espera: {espera} min | Urgencia: {urgencia:<5} | Premium: {str(premium):<5} => Acción: '{accion}' ({motivo})")
```
